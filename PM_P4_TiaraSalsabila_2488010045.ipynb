```python
import pandas as pd
import io

dummy_csv_data = """usia,pendapatan,pendidikan,kota,membeli
17,500000,SMA,Jogja,Tidak
24,750000,S1,Cirebon,Ya
28,900000,S2,Bandung,Tidak
18,700000,S1,Jogja,Ya
16,550000,SMA,Cirebon,Tidak
29,800000,S2,Bandung,Ya
"""

df = pd.read_csv(io.StringIO(dummy_csv_data))
df.head()
```

Langkah 1 — Menangani nilai hilang (imputasi median/modus).

```python
df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].mean())
```

Langkah 2 — Memisahkan fitur (X) dan label (y).

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
```

Langkah 3 — Encoding kategorikal (ordinal & one-hot).

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0, 'S1':1, 'S2':2})  # ordinal
X = pd.get_dummies(X, columns=['kota'], dtype=int)                # nominal
y = y.map({'Tidak':0, 'Ya':1})
```

Langkah 4 — Membagi data (SEBELUM scaling).

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42)
```

Langkah 5 — Penskalaan (fit pada latih, transform pada latih & uji).

```python
from sklearn.preprocessing import MinMaxScaler
num = ['usia', 'pendapatan', 'pendidikan']
sc = MinMaxScaler()
X_train[num] = sc.fit_transform(X_train[num])
X_test[num] = sc.transform(X_test[num])

display(X_train.describe())
```

Langkah 6 — Verifikasi: tidak ada nilai hilang, semua numerik, skala seragam.

```python
print("Missing values di X_train:\n", X_train.isnull().sum())

print("\nDeskripsi X_train:")
display(X_train.describe())
```

Jelaskan mengapa fit_transform pada latih tetapi hanya transform pada uji:

fit_transform pada latih agar komputer  belajar untuk menghitung rumus dari data latih, sedangkan agar kunci jawabannya tidak bocol ke soal atau data leakage pada data uji hanya transform saja yang mana hanya boleh numpang diukur menggunakan patokan data latih.
